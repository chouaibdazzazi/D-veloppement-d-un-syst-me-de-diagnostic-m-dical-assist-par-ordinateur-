```python
def Get_Transformer (isTrain):
    if isTrain :
        train_transformer = transforms.Compose([
        transforms.Resize((224,224)),
        transforms.RandomHorizontalFlip(0.5),
        transforms.RandomRotation(15),
        transforms.RandomVerticalFlip(0.5) , 
        transforms.ColorJitter(brightness=0.2 , contrast=0.2 , saturation=0.2),
        transforms.ToTensor(),
        transforms.Normalize(mean = [0.485, 0.456, 0.406],std = [0.229, 0.224, 0.225])
        ])
        return train_transformer
    
    else :
        test_transformer = transforms.Compose([
        transforms.Resize((224,224)),
        transforms.ToTensor(),
        transforms.Normalize(mean = [0.485, 0.456, 0.406],std = [0.229, 0.224, 0.225])
        ])
        return test_transformer
```

```python
def Get_labels_imgpaths(base_path ):
    
    image_paths = []
    labels = []
    
    if not os.path.isdir(base_path):
        raise ValueError(f"Base path {base_path} does not exist or is not a directory.")

    class_names = sorted([d for d in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, d))])
    if len(class_names) == 0:
        raise ValueError(f"No class subfolders found in {base_path}.")

    for idx, class_name in enumerate(class_names):
        class_folder = os.path.join(base_path, class_name)
        for img_path in glob(os.path.join(class_folder, "*")):
            image_paths.append(img_path)
            labels.append(idx)

    print(f"Found classes: {class_names}. Total images: {len(image_paths)}")
    return image_paths, labels, class_names

image_paths, labels, class_names = Get_labels_imgpaths(data_path)
num_classes = len(class_names)
print(f"Number of classes: {num_classes}")
```

```python
import shutil
from pathlib import Path

# Organize images from a single folder by inferring class from filename
def organize_from_single_folder(source_dir, target_base, class_names=None, copy_files=True):
    source_dir = Path(source_dir)
    target_base = Path(target_base)
    if not source_dir.exists():
        raise FileNotFoundError(f"Source folder not found: {source_dir}")
    # infer classes if not given
    if class_names is None:
        class_names = sorted([d.name for d in target_base.iterdir() if d.is_dir()])
    class_names_lower = [c.lower() for c in class_names]
    allowed = ('.jpg', '.jpeg', '.png', '.bmp')
    moved, ignored, ambiguous, errors = [], [], [], []
    for f in source_dir.iterdir():
        if not f.is_file() or not f.name.lower().endswith(allowed):
            continue
        name = f.name.lower()
        matches = [c for c in class_names_lower if c in name]
        if len(matches) == 1:
            dest = target_base / matches[0]
            dest.mkdir(parents=True, exist_ok=True)
            try:
                if copy_files:
                    shutil.copy2(f, dest / f.name)
                else:
                    shutil.move(str(f), str(dest / f.name))
                moved.append((str(f), matches[0]))
            except Exception as e:
                errors.append((str(f), str(e)))
        elif len(matches) == 0:
            ignored.append(str(f))
        else:
            ambiguous.append((str(f), matches))
    return {"moved": moved, "ignored": ignored, "ambiguous": ambiguous, "errors": errors}

# Simple verification to count images per class and show samples
from PIL import Image

def verify_dataset(base_path, max_samples=3):
    base = Path(base_path)
    if not base.exists():
        print(f"Base path not found: {base}")
        return False
    classes = sorted([d for d in base.iterdir() if d.is_dir()])
    if not classes:
        print(f"No class subfolders in {base}")
        return False
    print("Classes found:", [c.name for c in classes])
    total = 0
    for c in classes:
        files = [f for f in c.iterdir() if f.is_file() and f.suffix.lower() in ('.jpg','.jpeg','.png','.bmp')]
        print(f"  {c.name}: {len(files)} images")
        total += len(files)
        for f in files[:max_samples]:
            try:
                with Image.open(f) as im:
                    print(f"    sample: {f} size={im.size} mode={im.mode}")
            except Exception as e:
                print(f"    cannot open {f}: {e}")
    print(f"Total images: {total}")
    return True

# Run organizer with copy (safe)
source_dir = r"C:\Users\admin\Desktop\PFE\.ipynb_checkpoints"
print(f"Organizing images from: {source_dir} -> {data_path} (copy mode)")
report = organize_from_single_folder(source_dir, data_path, copy_files=True)
print(f"Moved: {len(report['moved'])} files")
print(f"Ignored (no match): {len(report['ignored'])}")
print(f"Ambiguous: {len(report['ambiguous'])}")
print(f"Errors: {len(report['errors'])}")
# show first 10 moved entries
for i,entry in enumerate(report['moved'][:10], 1):
    print(f"  {i}. {entry[0]} -> {entry[1]}")

# Copy all image files into an 'unknown' folder (if none matched earlier)
target_unknown = os.path.join(data_path, "unknown")
os.makedirs(target_unknown, exist_ok=True)
allowed = ('.jpg', '.jpeg', '.png', '.bmp')
copied = []
for f in Path(source_dir).iterdir():
    if f.is_file() and f.suffix.lower() in allowed:
        # only copy if not already copied/moved to existing class folders
        dest = Path(target_unknown) / f.name
        if not dest.exists():
            try:
                shutil.copy2(f, dest)
                copied.append(str(f))
            except Exception as e:
                print(f"Error copying {f}: {e}")
print(f"Copied {len(copied)} files to {target_unknown}")

# Run verification on target dataset
print('\nVerifying dataset after organizing and copying to unknown:')
verify_dataset(data_path)
```

```python
# Smoke test: dataset discovery, split, and short training sanity check
print('\n--- Smoke test: dataset discovery and short training run ---')
try:
    image_paths, labels, class_names = Get_labels_imgpaths(data_path)
except Exception as e:
    print('Dataset discovery failed:', e)
    raise

from collections import Counter
print('Class distribution:', Counter([class_names[l] for l in labels]))

if len(image_paths) == 0:
    print('No images found. Place images in extracted_images/<class>/ and re-run.')
else:
    try:
        train_image_paths, test_image_paths, train_labels, test_labels = train_test_split(
            image_paths, labels, random_state=seed, stratify=labels, test_size=0.2)
    except Exception as e:
        print('Stratified split failed:', e)
        print('Falling back to non-stratified split')
        train_image_paths, test_image_paths, train_labels, test_labels = train_test_split(
            image_paths, labels, random_state=seed, test_size=0.2)

    print(f'Train samples: {len(train_image_paths)}, Test samples: {len(test_image_paths)}')

    # small batch sizes for smoke test
    smoke_train_bs = min(8, max(1, len(train_image_paths)))
    smoke_test_bs = min(8, max(1, len(test_image_paths)))

    train_loader = Get_Dataloaders(train_image_paths, train_labels, smoke_train_bs, Get_Transformer(True), shuffle=True)
    test_loader = Get_Dataloaders(test_image_paths, test_labels, smoke_test_bs, Get_Transformer(False), shuffle=False)

    model = BrainXNet(num_classes=len(class_names)).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()

    # Train for 1 epoch, but limit batches
    model.train()
    max_train_batches = 10
    train_batches = 0
    total_loss = 0.0

    for inputs, lbls in train_loader:
        inputs, lbls = inputs.to(device), lbls.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, lbls)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        train_batches += 1
        if train_batches >= max_train_batches:
            break

    if train_batches > 0:
        print(f'Stopped after {train_batches} train batches. Avg train loss: {total_loss/train_batches:.4f}')
    else:
        print('No training batches were available (dataset too small).')

    # Quick eval on up to 5 batches
    model.eval()
    val_batches = 0
    val_loss = 0.0
    correct = 0
    total_v = 0
    with torch.no_grad():
        for inputs, lbls in test_loader:
            inputs, lbls = inputs.to(device), lbls.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, lbls)
            val_loss += loss.item()
            _, preds = torch.max(outputs, 1)
            correct += (preds == lbls).sum().item()
            total_v += lbls.size(0)
            val_batches += 1
            if val_batches >= 5:
                break

    if val_batches > 0:
        print(f'Validation over {total_v} samples — Avg loss: {val_loss/val_batches:.4f}, Acc: {100*correct/total_v:.2f}%')
    else:
        print('No validation batches were available.')

print('--- Smoke test complete ---')
```

```python
class BrainTumorDataset(Dataset):
    def __init__(self , image_paths , labels , transform = None):
        super().__init__()
        # validate and filter unreadable images up-front
        validated_paths, validated_labels = [], []
        for p, l in zip(image_paths, labels):
            try:
                with Image.open(p) as im:
                    im.verify()  # fast check that file is a valid image
                validated_paths.append(p)
                validated_labels.append(l)
            except Exception as e:
                print(f"Warning: Skipping unreadable image: {p} ({e})")
        self.image_path = validated_paths
        self.labels = validated_labels
        self.transform = transform
    
    def __len__ (self):
        return len(self.labels)
    
    def __getitem__(self, index):
        # return label as torch LongTensor and transformed image
        label = torch.tensor(self.labels[index], dtype=torch.long)
        img_path = self.image_path[index]
        try:
            image = Image.open(img_path).convert("RGB")
        except Exception as e:
            # fallback: create a black image and log warning
            print(f"Warning: Failed to open image {img_path}. Using blank image instead. ({e})")
            image = Image.new('RGB', (224, 224), (0, 0, 0))
        
        if self.transform :
            image = self.transform(image)
        
        return image, label

def Get_Dataloaders(image_paths, labels, batchsize, transformer, shuffle=False):
    data = BrainTumorDataset(image_paths, labels, transformer)
    dataloader = DataLoader(data, batch_size=batchsize, shuffle=shuffle)
    return dataloader

train_transformer = Get_Transformer(True)
test_transformer = Get_Transformer(False)

'train_loader' in globals() or print('train_loader will be created when splits are set')
```

```python
class ExtendedModelV2(nn.Module):
    def __init__(self, backbone , inputshape, num_classes=2):
        super(ExtendedModelV2 , self).__init__()
        self.backbone = backbone
        self.input = inputshape
        self.num_classes = num_classes
        
        # Input => [inputshape * 1 * 1]
        self.fc = nn.Sequential(
            nn.Linear(self.input , 1024),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            
            nn.Linear(1024 , num_classes)
        )
        self.Avgpool = nn.AdaptiveAvgPool2d((1,1))
        
    def forward(self , x):
        x = self.backbone(x)
        x = self.Avgpool(x)
        x = x.view(x.size(0) , -1)
        x = self.fc(x)
        return x

# class ExtendedModel(nn.Module):
#     def __init__(self, backbone):
#         super(ExtendedModel , self).__init__()
#         self.backbone = backbone
#         # Input => [2048 * 1 * 1]
#         self.fc = nn.Sequential(
#             nn.Linear(2048 , 1024),
#             nn.ReLU(inplace=True),
#             nn.Dropout(0.5),
            
#             nn.Linear(1024 , 2)
#         )
#     def forward(self , x):
#         x = self.backbone(x)
#         x = x.view(x.size(0) , -1)
#         x = self.fc(x)
#         return x
```

```python
def Get_resnet152_backbone():
    orignal_resnet = models.resnet152(weights=models.ResNet152_Weights.IMAGENET1K_V1)
    backbone_model = list(orignal_resnet.children())[:-1] # Expected Output => [2048 * w * h]
    backbone_model = nn.Sequential(*backbone_model) 
    for param in backbone_model.parameters():
        param.requires_grad = False
    
    x = torch.randn(1,3,224,224)
    y = backbone_model(x)
    Output_shape = y.shape 
    
    return backbone_model , Output_shape


def Get_Efficientnet_backbone():
    efficientnet_b0_orignal=models.efficientnet_b0(weights = models.EfficientNet_B0_Weights.IMAGENET1K_V1)
    backbone = nn.Sequential(*list(efficientnet_b0_orignal.children())[:-1])
    
    for param in backbone.parameters():
        param.requires_grad = False
        
    x = torch.randn(1,3,224,224)
    y = backbone(x)
    Output_shape = y.size(1)
    return backbone , Output_shape
```

```python
class BrainXNet(nn.Module):
    def __init__(self, num_classes=2):
        super(BrainXNet , self).__init__()
        self.Feature = nn.Sequential(
            # Input = [3 * 224 * 224 ]
            
            nn.Conv2d(3,32,kernel_size=3 , padding='same'),
            nn.BatchNorm2d(32),
            # Output = [32 * 224 * 224]
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size= 2 , stride= 2),
            # Output = [32 * 112 * 112]
            
            nn.Conv2d(32 , 64 , kernel_size= 3 , padding='same'),
            nn.BatchNorm2d(64),
            # Output = [64 * 112 * 112]
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size= 2 , stride= 2),
            # Output = [64 * 56 * 56]
            
            nn.Conv2d(64 , 128 , kernel_size= 3 , padding='same'),
            nn.BatchNorm2d(128),
            # Output = [128 * 56 * 56]
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size= 2 , stride= 2),
            # Output = [128 * 28 * 28]
            
            nn.Conv2d( 128 ,256, kernel_size= 3 , padding='same'),
            nn.BatchNorm2d(256),
            # Output = [256 * 28 * 28]
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size= 2 , stride= 2),
            # Output = [256 * 14 * 14]
            
            nn.Conv2d(256 , 512 , kernel_size= 3 , padding="same"),
            nn.BatchNorm2d(512),
            # Output = [512 * 14 * 14]
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size= 2 , stride= 2),
            # Output = [512 * 7 * 7]
            
            nn.Conv2d(512 , 1024 , kernel_size= 3 , padding="same"),
            nn.BatchNorm2d(1024),
            # Output = [1024 * 14 * 14]
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size= 2 , stride= 2)
            # Output = [1024 * 7 * 7]
        )
        self.Avgpool = nn.AdaptiveAvgPool2d((1,1))
        
        self.classifier = nn.Sequential(
            nn.Linear(1024 * 1 * 1 , 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5) ,
            
            nn.Linear(512 , num_classes)
        )
        self.num_classes = num_classes
    
    def forward (self , x):
        x = self.Feature(x)
        x = self.Avgpool(x)
        # print("Shape before flatten:", x.shape) 
        x = x.view(x.size(0), -1)
        # print("Shape after flatten:", x.shape)
        x = self.classifier(x)
        return x
```

```python
def Train_Model(model ,train_loader , optimizer , criterion ,device ):
    
    model.train()
    total_loss, correct, total = 0.0, 0.0, 0.0
    
    for inputs, labels in tqdm(train_loader, desc="Training", leave=False):
        inputs, labels = inputs.to(device), labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs , labels)
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
        _,predicted = torch.max(outputs , 1)
        correct += (predicted == labels).sum().item()
        total += labels.size(0)
        
    Train_loss = total_loss / len(train_loader)
    Train_acc = 100 * correct / total
    
    return Train_loss , Train_acc
```

```python
def Evaluate_model (model, test_loader, criterion,device):
    model.eval()
    total_loss, correct, total = 0.0, 0.0, 0.0
    
    with torch.no_grad():
        for inputs , labels in tqdm(test_loader, desc="Evaluation", leave=False):
            inputs, labels = inputs.to(device), labels.to(device)
            
            outputs = model(inputs)
            
            loss = criterion(outputs , labels)
            total_loss += loss.item()
            _,predicted = torch.max(outputs , 1)
            correct += (predicted == labels).sum().item()
            total += labels.size(0)
        
        Validation_loss = total_loss / len(test_loader)
        Validation_acc = 100 * correct / total
        
        return Validation_loss , Validation_acc
```

```python


def plot_curves(train_losses, val_losses, train_accuracies, val_accuracies, model_name="Model"):
    """
    Plot and save training & validation loss and accuracy curves.
    Works both locally and on Kaggle.
    """
    results_dir = os.path.join(".", "Results")
    os.makedirs(results_dir, exist_ok=True)
    save_path = os.path.join(results_dir, f"{model_name}_Training_Results.png")

    epochs = range(1, len(train_losses) + 1)
    plt.figure(figsize=(12, 5))

    # ---- Loss Plot ----
    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_losses, 'b', label='Training Loss')
    plt.plot(epochs, val_losses, 'r', label='Validation Loss')
    plt.title(f'{model_name}: Training vs Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    # ---- Accuracy Plot ----
    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_accuracies, 'b', label='Training Accuracy')
    plt.plot(epochs, val_accuracies, 'r', label='Validation Accuracy')
    plt.title(f'{model_name}: Training vs Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()

    print(f"Plot saved successfully at: {save_path}")
```

```python
# Recreate dataset split and model to ensure sizes match before full training
print('\n--- Preparing for training: verify dataset & model sizes ---')
image_paths, labels, class_names = Get_labels_imgpaths(data_path)
num_classes = len(class_names)
from collections import Counter
print('Class distribution:', Counter([class_names[l] for l in labels]))
print(f'num_classes = {num_classes}')

# Safely create train/test splits
try:
    train_image_paths, test_image_paths, train_labels, test_labels = train_test_split(
        image_paths, labels, random_state=seed, stratify=labels, test_size=0.2)
except Exception as e:
    print('Stratified split failed, falling back to non-stratified split:', e)
    train_image_paths, test_image_paths, train_labels, test_labels = train_test_split(
        image_paths, labels, random_state=seed, test_size=0.2)

print(f'Train samples: {len(train_image_paths)}, Test samples: {len(test_image_paths)}')

# Build dataloaders (use configured batch sizes)
train_loader = Get_Dataloaders(train_image_paths, train_labels, train_batch, Get_Transformer(True), shuffle=True)
test_loader = Get_Dataloaders(test_image_paths, test_labels, test_batch, Get_Transformer(False), shuffle=False)

# Recreate model with the correct number of outputs
model = BrainXNet(num_classes=num_classes).to(device)
# Alternative models (uncomment to use):
# model = ExtendedModelV2(efficientnet_b0_Backbone , input_shape, num_classes=num_classes).to(device)

save_dir = os.path.join(os.getcwd(), "Saved_Models")
os.makedirs(save_dir, exist_ok=True)
best_model_path = os.path.join(save_dir, f"{model.__class__.__name__}_best.pt")

optimizer = torch.optim.AdamW(model.parameters() , lr=lr)
criterion = nn.CrossEntropyLoss()
train_losess , val_losess , train_acces , val_acces = [],[],[],[]
best_val_loss = float('inf')

# Sanity check: ensure labels are in range
if labels:
    max_label = max(labels)
else:
    max_label = -1
if max_label >= num_classes:
    raise ValueError(f"Label value {max_label} >= num_classes {num_classes} — recreate model with more outputs or fix labels.")

# Training loop with diagnostic on IndexError
try:
    for epoch in range(epochs):
        train_loss, train_acc = Train_Model(model, train_loader, optimizer, criterion, device)
        val_loss, val_acc = Evaluate_model(model, test_loader, criterion, device)
        
        train_losess.append(train_loss)
        train_acces.append(train_acc)
        val_losess.append(val_loss)
        val_acces.append(val_acc)
        
        print(f"Epoch [{epoch+1:02d}/{epochs}] "
              f"| Train Loss: {train_loss:.4f} | Train Acc: {train_acc:6.2f}% || "
              f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:6.2f}%")
        
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), best_model_path)
            print(f"Best model saved at Epoch [{epoch+1}] with Val Acc: {val_acc:.4f}")
except IndexError as e:
    # Provide debugging info to help diagnose label/model mismatch
    print('IndexError during training — likely label out of range for model outputs.')
    print('Exception:', e)
    # show example batch diagnostic (if possible)
    try:
        batch_inputs, batch_labels = next(iter(train_loader))
        print('Batch shapes — inputs:', getattr(batch_inputs, 'shape', None), 'labels:', getattr(batch_labels, 'shape', None))
        print('Unique labels in batch:', torch.unique(batch_labels))
        print('Model output size for the batch (before softmax):', model(batch_inputs.to(device)).shape)
    except Exception as inner_e:
        print('Could not run diagnostic batch:', inner_e)
    raise
```

```python
# Short controlled smoke test — 1 epoch, limited batches (safe interrupt)
print('\n--- Short smoke test (1 epoch, limited batches) ---')
image_paths, labels, class_names = Get_labels_imgpaths(data_path)
num_classes = len(class_names)
from collections import Counter
print('Class distribution:', Counter([class_names[l] for l in labels]))

if len(image_paths) == 0:
    print('No images available. Add images and re-run.')
else:
    try:
        train_image_paths, test_image_paths, train_labels, test_labels = train_test_split(
            image_paths, labels, random_state=seed, stratify=labels, test_size=0.2)
    except Exception as e:
        print('Stratified split failed, falling back to non-stratified split:', e)
        train_image_paths, test_image_paths, train_labels, test_labels = train_test_split(
            image_paths, labels, random_state=seed, test_size=0.2)

    print(f'Train samples: {len(train_image_paths)}, Test samples: {len(test_image_paths)}')

    smoke_train_bs = min(4, max(1, len(train_image_paths)))
    smoke_test_bs = min(4, max(1, len(test_image_paths)))

    train_loader = Get_Dataloaders(train_image_paths, train_labels, smoke_train_bs, Get_Transformer(True), shuffle=True)
    test_loader = Get_Dataloaders(test_image_paths, test_labels, smoke_test_bs, Get_Transformer(False), shuffle=False)

    model = BrainXNet(num_classes=num_classes).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()

    model.train()
    max_train_batches = 5
    t_batches = 0
    t_loss = 0.0
    t_correct = 0
    t_total = 0

    for inputs, lbls in train_loader:
        inputs, lbls = inputs.to(device), lbls.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, lbls)
        loss.backward()
        optimizer.step()

        t_loss += loss.item()
        _, preds = torch.max(outputs, 1)
        t_correct += (preds == lbls).sum().item()
        t_total += lbls.size(0)
        t_batches += 1
        if t_batches >= max_train_batches:
            break

    if t_batches > 0:
        print(f'Training — batches: {t_batches}, avg_loss: {t_loss/t_batches:.4f}, acc: {100*t_correct/t_total:.2f}%')
    else:
        print('No train batches available.')

    # quick eval
    model.eval()
    v_batches = 0
    v_loss = 0.0
    v_correct = 0
    v_total = 0
    with torch.no_grad():
        for inputs, lbls in test_loader:
            inputs, lbls = inputs.to(device), lbls.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, lbls)
            v_loss += loss.item()
            _, preds = torch.max(outputs, 1)
            v_correct += (preds == lbls).sum().item()
            v_total += lbls.size(0)
            v_batches += 1
            if v_batches >= 2:
                break

    if v_batches > 0:
        print(f'Validation — batches: {v_batches}, avg_loss: {v_loss/v_batches:.4f}, acc: {100*v_correct/v_total:.2f}%')
    else:
        print('No validation batches available.')

print('--- Short smoke test complete ---')
```
